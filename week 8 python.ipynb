```python
# Q9. Factorial using Function

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
```

```python
# Q10. Prime Number Checker

def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
# Q15. Word Frequency Counter

sentence = input("Enter sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
# Q16. Library Book System

books = []

def add_book(book):
    books.append(book)

def remove_book(book):
    if book in books:
        books.remove(book)

def search_book(book):
    if book in books:
        print("Book Found")
    else:
        print("Book Not Found")

def display_books():
    print(books)

add_book("Python")
add_book("Java")
remove_book("Java")
search_book("Python")
display_books()
```

```python
# Q17. Student Result Analyzer

marks = [35, 75, 90, 45, 60]

def average_marks(marks):
    return sum(marks) / len(marks)

def highest_marks(marks):
    return max(marks)

def lowest_marks(marks):
    return min(marks)

def passed_students(marks):
    count = 0
    for mark in marks:
        if mark > 40:
            count += 1
    return count

print("Average:", average_marks(marks))
print("Highest:", highest_marks(marks))
print("Lowest:", lowest_marks(marks))
print("Passed Students:", passed_students(marks))
```

```python
# Q18. Password Strength Checker

def check_password(password):

    if (len(password) >= 8 and
        any(char.isdigit() for char in password) and
        any(char.isupper() for char in password) and
        any(char.islower() for char in password)):

        return "Strong Password"

    return "Weak Password"

pwd = input("Enter password: ")
print(check_password(pwd))
```

```python
# Q19. Contact Book Application

contacts = {}

def add_contact(name, number):
    contacts[name] = number

def search_contact(name):
    if name in contacts:
        print(name, contacts[name])
    else:
        print("Contact not found")

def delete_contact(name):
    if name in contacts:
        del contacts[name]

def display_contacts():
    print(contacts)

add_contact("Rahul", 9876543210)
search_contact("Rahul")
delete_contact("Rahul")
display_contacts()
```

```python
# Q20. Mini Banking System

balance = 1000

def deposit(amount):
    global balance
    balance += amount

def withdraw(amount):
    global balance
    if amount <= balance:
        balance -= amount
    else:
        print("Insufficient Balance")

def check_balance():
    print("Balance:", balance)

deposit(500)
withdraw(300)
check_balance()
```
